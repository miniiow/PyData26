## SECTION 01 결측치 처리

```python
import pandas as pd
url = 'https://raw.githubusercontent.com/sehakflower/data/main/titanic.csv'
titanic_df = pd.read_csv(url, sep='\t')
titanic_df.info()
```

```python
# 모든 열을 소문자로 변경

new_columns=['passengerId', 'servived', 'pclass', 'name', 'sex', 'age', 'sibsp', 'parch', 'ticket', 'fare', 'cabin', 'embarked']
titanic_df.columns = new_columns
titanic_df.columns
```

### 결측치 확인 함수

```python
# isna 함수

titanic_df['age'][titanic_df['age'].isna()]
```

```python
# isnull 함수

titanic_df['age'].isnull()
titanic_df['age'][titanic_df['age'].isnull()]
titanic_df['age'][titanic_df['age'].isnull()].isnull().sum()
```

```python
# notnull 함수

titanic_df['age'].notnull()
titanic_df['age'][titanic_df['age'].notnull()]
titanic_df.loc[titanic_df['age'].notnull(), ['age', 'name']]
titanic_df.loc[titanic_df['age'].isnull(), ['age','name']]
```

### 결측치 처리방법

```python
# 결측치를 다른 문자로 변경하기

titanic_df.head()
titanic_df['cabin']=titanic_df['cabin'].fillna('N')
titanic_df['cabin']

titanic_df['cabin'].value_counts()
titanic_df['cabin'][0]

titanic_df['cabin']=titanic_df['cabin'].apply(lambda x: x[0])
titanic_df['cabin']
titanic_df.info()
```

```python
# 결측치를 평균값으로 변경하기

titanic_df_copy = titanic_df.copy()
titanic_df_copy
```

```python
titanic_df_copy['age'].isnull()
```

```python
titanic_df_copy['age'] = titanic_df_copy['age'].fillna(titanic_df_copy['age'].mean())
titanic_df_copy['age']
```

```python
# 데이터프레임 복사 - 얕은 복사와 깊은 복사

copy_titanic_df = titanic_df.copy()
copy_titanic_df
```

```python
# hex함수를 이용해 원본과 복사된 데이터프레임의 주소를 확인

hex(id(titanic_df))
```

```python
hex(id(copy_titanic_df))
```

```python
# head함수를 이용해 원본인 titanic_df와 복사본인 copy_titanic_df의 내용을 확인

titanic_df.head()
```

```python
copy_titanic_df.head()
```

```python
# copy_titanic_df의 age열을 모두 0으로 변경하고 head 함수로 확인

copy_titanic_df['age'] = 0
copy_titanic_df.head()
```

```python
# copy함수로 복사면 원본데이터프레임의 값이 변경되지 않는 것을 확인 >> ㄷㅏ른 주소값이라서

titanic_df.head()
```
