## 1. 파이썬 파일 처리

```python
sample_file = 'data/sample.txt'
```

```python
# 파일을 open하여 전체 내용을 읽어서 출력
with open(sample_file, mode='rt', encoding='utf-8') as f:
    content = f.read()
    print(content)
print("----- 끝 -----")
```

```python
# 파일을 open하고 한 줄씩 읽어서 출력하기
with open(sample_file, 'r', encoding='utf-8') as f:
    lines = f.readlines()
    for line in lines:
        print(line, end="")
print("----- 끝 -----")
```

```python
with open(sample_file, 'r', encoding='utf-8') as f:
    line = f.readline()
    while line:
        print(line, end="")
        line = f.readline()
```

```python
# 파일객체에서 한 줄씩 읽어 오기
# 파일 객체 f를 반복문에서 사용하면 파일을 한 줄씩 읽으며,
# 각 줄 끝의 \n도 포함해서 가져옵
with open(sample_file, 'r', encoding='utf-8') as f:
    for line in f:
        print(line, end='')

print("----- 끝 -----")
```

```python
# 여러 개의 파일 읽기 (모든 txt 파일을 읽어서 출력하기)
import glob
txt_file_pattern = "data/*.txt"
files = glob.glob(txt_file_pattern)
print(files)

for file in files:
    print(file)

    with open(file, mode = 'r', encoding='utf-8') as f:
            line = f.read()
            print(line, end='')

print("----- 끝 -----")
```

```python
# 지역별 총매출

sales_df.groupby(by='지역')['매출'].sum()
```

```python
# 상품별 평균 매출 > 상품을 기준으로 매출의 평균 계산

sales_df.groupby(by='상품')['매출'].mean()
```

```python
# 지역별 판매수량의 합계와 매출의 합계 > 지역을 기준으로 판매수량과 매출의 각각 합계 구하기

sales_df.groupby(by='지역')[['판매수량', '매출']].sum()
```

```python
# 지역과 상품별 평균 매출

sales_df.groupby(by=['지역', '상품'])['매출'].mean()
```

```python
# 위의 실습을 pivot_table()로 적용

sales_df.pivot_table(
    values='매출',
    aggfunc='mean',
    index='지역',
    columns='상품'
)
```

```python
# 지역별·상품별 총매출을 구하고 행과 열의 합계 표시

sales_df.pivot_table(
    index='상품',
    columns='지역',
    values='매출',
    aggfunc='sum',
    margins=True    # 전체를 통합한 것
)
```

```python
# 지역별 매출 : 건수, 합계, 평균, 최대값을 한번에 출력

sales_df.groupby(by='지역')['매출'].agg(['count', 'sum', 'mean', 'max'])
```
